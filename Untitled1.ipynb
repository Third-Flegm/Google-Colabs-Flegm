```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
import tiktoken

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Training on:", device)

# --- Upgraded Hyperparameters ---
FILE_NAME = "books_set.txt"#"big_set.txt"   # Change to "set1.txt" or your dataset filename
CONTEXT_LENGTH = 64         # Looks at 64 tokens (~45-50 words) of history
EMBED_DIM = 128             # Richer internal representation per token
N_HEAD = 4                  # 4 attention heads for grammar & context
N_LAYERS = 4                # 4 Transformer blocks deep
FF_DIM = 512                # Feedforward hidden size
DROPOUT = 0.1               # Prevents memorization
BATCH_SIZE = 64             # Safe for T4 GPU memory
EPOCHS = 10                 # On millions of words, 3-10 epochs is plenty
LEARNING_RATE = 3e-4        # Standard Transformer learning rate

# 1. Load Sub-word Tokenizer (GPT-2 BPE vocabulary = 50,257 tokens)
enc = tiktoken.get_encoding("gpt2")
vocab_size = enc.n_vocab

# 2. Load and tokenize the dataset
print(f"Reading and tokenizing {FILE_NAME}...")
with open(FILE_NAME, "r", encoding="utf-8") as f:
    text = f.read()

tokens = enc.encode(text)
data = torch.tensor(tokens, dtype=torch.long)
print(f"Total tokens: {len(data):,} | Fixed Vocab Size: {vocab_size:,}")

# 3. Create training chunks fast using tensor unfolding (low RAM usage)
# Stride = CONTEXT_LENGTH // 2 avoids redundant overlapping copies on huge files
stride = CONTEXT_LENGTH // 2
chunks = data.unfold(0, CONTEXT_LENGTH + 1, stride)
X = chunks[:, :-1]
Y = chunks[:, 1:]

dataset = TensorDataset(X, Y)
loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, pin_memory=True)
print(f"Training batches per epoch: {len(loader):,}")

# 4. Define Upgraded TinySLM
class TinySLM(nn.Module):
    def __init__(self, vocab_size, embed_dim, context_len, n_head, n_layers, ff_dim, dropout):
        super().__init__()
        self.token_emb = nn.Embedding(vocab_size, embed_dim)
        self.pos_emb = nn.Embedding(context_len, embed_dim)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim,
            nhead=n_head,
            dim_feedforward=ff_dim,
            dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)
        self.fc_out = nn.Linear(embed_dim, vocab_size)

    def forward(self, x):
        B, T = x.shape
        positions = torch.arange(T, device=x.device)
        x = self.token_emb(x) + self.pos_emb(positions)
        mask = nn.Transformer.generate_square_subsequent_mask(T).to(x.device)
        x = self.transformer(x, mask=mask, is_causal=True)
        return self.fc_out(x)

# 5. Initialize Model, Optimizer & Mixed Precision Scaler (2x faster on T4 GPU)
model = TinySLM(
    vocab_size, EMBED_DIM, CONTEXT_LENGTH, N_HEAD, N_LAYERS, FF_DIM, DROPOUT
).to(device)

optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE)
loss_fn = nn.CrossEntropyLoss()
scaler = torch.amp.GradScaler("cuda", enabled=(device == "cuda"))

param_count = sum(p.numel() for p in model.parameters())
print(f"Model parameters: {param_count:,}")

# 6. Training Loop with Auto-Save every epoch
for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for step, (batch_x, batch_y) in enumerate(loader):
        batch_x = batch_x.to(device, non_blocking=True)
        batch_y = batch_y.to(device, non_blocking=True)

        optimizer.zero_grad()
        with torch.amp.autocast("cuda", enabled=(device == "cuda")):
            logits = model(batch_x)
            loss = loss_fn(logits.view(-1, vocab_size), batch_y.view(-1))

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item()

        if (step + 1) % 500 == 0:
            print(f"  Epoch {epoch + 1} | Step {step + 1}/{len(loader)} | Loss: {loss.item():.4f}")

    avg_loss = total_loss / len(loader)
    print(f"==> Epoch {epoch + 1}/{EPOCHS} Complete | Avg Loss: {avg_loss:.4f}")

    # Save checkpoint after every epoch so you never lose progress
    torch.save({
        "model_state": model.state_dict(),
        "tokenizer": "gpt2",
        "context_length": CONTEXT_LENGTH,
        "embed_dim": EMBED_DIM,
        "n_head": N_HEAD,
        "n_layers": N_LAYERS,
        "ff_dim": FF_DIM
    }, "brain.pt")
    print("    Saved checkpoint to brain.pt")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import tiktoken

device = "cuda" if torch.cuda.is_available() else "cpu"

class TinySLM(nn.Module):
    def __init__(self, vocab_size, embed_dim, context_len, n_head, n_layers, ff_dim, dropout=0.0):
        super().__init__()
        self.token_emb = nn.Embedding(vocab_size, embed_dim)
        self.pos_emb = nn.Embedding(context_len, embed_dim)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim, nhead=n_head, dim_feedforward=ff_dim, dropout=dropout, batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)
        self.fc_out = nn.Linear(embed_dim, vocab_size)

    def forward(self, x):
        B, T = x.shape
        positions = torch.arange(T, device=x.device)
        x = self.token_emb(x) + self.pos_emb(positions)
        mask = nn.Transformer.generate_square_subsequent_mask(T).to(x.device)
        x = self.transformer(x, mask=mask, is_causal=True)
        return self.fc_out(x)

# 1. Load checkpoint and tokenizer
checkpoint = torch.load("brain.pt", map_location=device, weights_only=True)
enc = tiktoken.get_encoding(checkpoint["tokenizer"])
context_length = checkpoint["context_length"]

model = TinySLM(
    enc.n_vocab,
    checkpoint["embed_dim"],
    context_length,
    checkpoint["n_head"],
    checkpoint["n_layers"],
    checkpoint["ff_dim"]
).to(device)
model.load_state_dict(checkpoint["model_state"])
model.eval()

# 2. Start from a prompt
prompt = input("Enter starting prompt (e.g., 'Es war einmal' or 'Der Sinn des Lebens'): ")
if not prompt.strip():
    prompt = "Die"

token_ids = enc.encode(prompt)

TEMPERATURE = 0.72
TOP_K = 30
REPETITION_PENALTY = 1.15  # Bestraft Wörter/Tokens, die gerade erst benutzt wurden

with torch.no_grad():
    for _ in range(300):
        context = torch.tensor([token_ids[-context_length:]], dtype=torch.long, device=device)
        logits = model(context)[0, -1, :]

        # 1. Repetition Penalty auf die letzten 30 Tokens anwenden
        for prev_id in set(token_ids[-30:]):
            if logits[prev_id] > 0:
                logits[prev_id] /= REPETITION_PENALTY
            else:
                logits[prev_id] *= REPETITION_PENALTY

        logits = logits / TEMPERATURE

        # 2. Top-K Sampling
        top_vals, top_idx = torch.topk(logits, TOP_K)
        probs = F.softmax(top_vals, dim=-1)
        next_token = top_idx[torch.multinomial(probs, num_samples=1)].item()

        token_ids.append(next_token)


generated_text = enc.decode(token_ids)
print("\n--- GENERATED TEXT ---\n")
print(generated_text)

with open("output.txt", "w", encoding="utf-8") as f:
    f.write(generated_text)
```
